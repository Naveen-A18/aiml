```python
import tkinter
from tkinter import *

root = Tk()
root.title("calculator")
root.config(bg="gray")
root.geometry("800x1200")

def add():
    print("adding your number")
    num1 = int(number1.get())
    num2 = int(number2.get())
    result = num1 + num2
    outputText.delete("1.0", END)
    outputText.insert(END, f"Addition Result: {result}\n")

def sub():
    print("subtracting your number")
    num1 = int(number1.get())
    num2 = int(number2.get())
    result = num1 - num2
    outputText.delete("1.0", END)
    outputText.insert(END, f"Subtracting Result: {result}\n")

def multi():
    print("multiplying your number")
    num1 = int(number1.get())
    num2 = int(number2.get())
    result = num1 * num2
    outputText.delete("1.0", END)
    outputText.insert(END, f"Multiplication Result: {result}\n")

def div():
    print("dividing your number")
    num1 = int(number1.get())
    num2 = int(number2.get())
    if num2 == 0:
        outputText.delete("1.0", END)
        outputText.insert(END, "Error: Cannot divide by zero")
    else:
        result = num1 / num2
        outputText.delete("1.0", END)
        outputText.insert(END, f"Division Result: {result}\n")

number1 = StringVar()
number2 = StringVar()

l0=Label(root,text="CALCULATOR",bg="pink",fg="blue",font="times 20 bold")
l0.pack()

l1 = Label(root, text="Enter number 1", font="times 20 bold", bg="pink").place(x=40, y=50)
l2 = Label(root, text="Enter number 2", font="times 20 bold", bg="pink").place(x=40, y=110)
l3 = Label(root, text="Output : ",font="times 20 bold", bg="pink").place(x=40,y=250)

entrynum1 = Entry(root, font="times 20 bold",bg="pink",fg="blue", textvariable=number1)
entrynum1.place(x=310, y=50)

entrynum2 = Entry(root, font="times 20 bold",bg="pink",fg="blue", textvariable=number2)
entrynum2.place(x=310, y=110)

outputText = Text(root, height=15, width=40,bg="pink",fg="blue",font="times 13 bold")
outputText.place(x=240, y=300,)

Button(root, text="+", font="elephant 14 bold",bg="pink", command=add).place(x=40, y=170)
Button(root, text="-", font="elephant 14 bold",bg="pink", command=sub).place(x=120, y=170)
Button(root, text="*", font="elephant 14 bold",bg="pink", command=multi).place(x=190, y=170)
Button(root, text="/", font="elephant 14 bold",bg="pink", command=div).place(x=270, y=170)
Button(root, text="Exit", font="elephant 14 bold",bg="pink",fg="blue", command=root.destroy).place(x=510, y=170)

root.mainloop()
```
