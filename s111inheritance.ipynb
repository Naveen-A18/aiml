```python
#single inheritance example
class parentclass: #parent class
    def myfunction(self):
        print("parent class attribute called")
class childclass(parentclass):
    def myfunction2(self):
        print("child class attribute called")
c1 = parentclass()
c1.myfunction()
c2=childclass()
c2.myfunction2()
```

```python
#multilevel inheritance
class A: #parent class
    def myfunction(self):
        print("class A funtion is called")
class B(A):
    def myfunction2(se):
        print("class b function is called")
class C(B):
    def myfunction3(self):
        print("class c function is called")
c1 = C()
c1.myfunction3()
c1.myfunction2()
c1.myfunction()
```

```python
# Multiple inheritance
class A1: # 1st parent class >mamma
    def myfunction(self):
        print("Your mamma properties get")
class B1: # 2nd parent class > papa
    def myfunction2(self):
        print("Your father properties get")
class C(A1,B1):
    def myfunction3(self):
        print("I have access to get the property of both")
child = C()
child.myfunction3()
child.myfunction2()
child.myfunction()
```
